```python
Q = "Press Q = Exit"
M = "Press M = Main"
S = "Press S = Start in Program"
state = True
while(state):
    print(Q)
    print(M)
    print(S)

    inp = input("Menu Selected : ")
    if(inp == 'Q'):
        state = False
    elif(inp == 'M'):
        print("Home Menu")
    elif(inp == 'S'):
        try:
            #ให้ผู้ใช้กรอกจำนวนรอบที่ต้องการ
            rounds_str = input("กรุณาระบุจำนวนรอบ (สูงสุด 10 รอบ): ")
            rounds = int(rounds_str)
            
            #ตรวจสอบเงื่อนไข: ถ้ากรอกมาเกิน 10 ให้ปรับเหลือแค่ 10 รอบเท่านั้น
            if rounds > 10:
                print(f"คุณใส่มา {rounds} รอบ เกินลิมิต! กรุณาปรับเหลือ 10 รอบ")
                rounds = 10
            
            print(f"เริ่มวิ่งกันเลย ")
            
            #เริ่มวิ่งตามจำนวน rounds ที่ผ่านการตรวจสอบแล้ว
            for i in range(1, rounds + 1):
                print(f"วิ่งรอบที่ : {i}")
            
            print("วิ่งครบแล้ว กลับไปหน้าเมนูกันเลย")
            
        except ValueError:
            print("เกิดข้อผิดพลาด! กรุณาใส่ตัวเลขเท่านั้น")
    else:
        print("Home Menu")
```
